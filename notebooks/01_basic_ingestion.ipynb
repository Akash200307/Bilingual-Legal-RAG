```python
from src.ingestion import load_documents,chunking,pdfs
from src.Chromadb_init import build_chroma_index
from src.embeddings import get_minilm_embed
```

```python
docs = load_documents(pdfs, num_files_limit=80)


nodes = chunking(docs)

```

```python
import time
embed = get_minilm_embed()
t0 = time.time()
index = build_chroma_index(
    nodes,
    persis_dir="../indexes/chroma_minilm",
    embed_model=embed,
    collection_name="legal_minilm"
)
print(f"Indexing time: {(time.time()-t0)/60:.1f} min")
```

```python
from src.retreival import make_query_engine
import  time
query_engine=make_query_engine(index,top_n=5)

questions=[
    "what category cases mostly occurred in 1975?",
    "How many civil and criminal cases are solved? ",
    "Who is the most successful judge gave decision in minimal time?",
    "How much money laundering have happened and money value too?"
]
results=[]
for q in questions:
    t0=time.time()
    ans=query_engine.query(q)
    latency=time.time()-t0
    results.append({"questions":q,"answer":ans,"Latency":latency})

import pandas as pd

pd.DataFrame(results).to_csv("../results/week1_qa.csv",index=False)
```
